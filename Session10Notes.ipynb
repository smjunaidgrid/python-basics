```python
try:
    f = open('testfile','')
    f.write("Write a test line")
except SyntaxError:
    print("Check with the Syntax")
except AttributeError:
    print("There is an attribute error")
except NameError:
    print("NAME ERROR!")
except SystemError:
    print("Check with the sytax")
except TypeError:
    print("There was a type error")
except OSError:
    print("There is a OS Error")
except:
    print("EXCEPTION")
finally:
    print("Finally Running")
```
