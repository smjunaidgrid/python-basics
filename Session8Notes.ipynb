```python
class Cat():
    def __init__(self,mybreed,name,spots):

        # Attribute - characteristic of the object 
        self.breed = mybreed 
        # parameter is assigned to an object that could be later called on 

        self.name = name

        self.spots = spots
```

```python
my_cat = Cat(mybreed='British',name='Zoey',spots = False)
```

```python
class Cat():
    # Class object attribute - Same for the any instance of the class 
    species = 'mammal'
    def __init__(self,mybreed,name,spots):

        # Attribute - characteristic of the object 
        self.breed = mybreed 
        # parameter is assigned to an object that could be later called on 

        self.name = name

        self.spots = spots
    def meows(self):
        print("MEOWW!")
```

```python
my_cat = Cat(mybreed='British',name='Zoey',spots = False)
```

```python
my_cat.species
```

```python
# Methods - Functions defined inside the body of the class to define object of the class through self arg or keyword
# Method - Function of class that will work the object someway 
my_cat.meows()
```
