```python
def hello(name = "Junaid"):
    print("The hello() function is been executed")

    def greet():
        print("This is a greet function")
    
    def welcome():
        print("This is a welcome() function")
    
    if name == "Junaid":
        return greet
    else:
        return welcome

    print("This is the end of hello func")
```

```python
newfunc = hello("Junaid")
```

```python
print(newfunc())
```

```python
newfunc()
```

```python
def new_decorator(orgfunc):
    def wrapfunc():
        print("Code Beforee Wrap func")
        orgfunc()
        print("Code After new func")
    return wrapfunc
```

```python
def func_dec():
    print("Decorated Func")
    
```

```python
func_dec()
```

```python
decorated_func = new_decorator(func_dec)
```

```python
decorated_func()
```

```python
@new_decorator
def func_dec():
    print("Decorated Func")
    
```

```python
func_dec()
```

```python
def create_squares(n):
    for x in range(n):
        yield x**2
```

```python
list(create_squares(10))
    
```

```python
for x in create_squares(5):
    print(x)
```

```python
def fib(n):
    a = 1
    b = 1 
    for i in range(n):
        yield a
        a,b = b,a+b
```

```python
for num in fib(10):
    print(num)
```
