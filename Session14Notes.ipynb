```python
def func_one(n):
    return [str(num) for num in range(n)]

```

```python
func_one(10)
```

```python
def func_two(n):
    return list(map(str,range(n)))
```

```python
func_two(10)
```

```python
import time 
start_time = time.time()
res = func_one(100000000)
end_time = time.time()
elapsed_time = end_time - start_time
 
print(elapsed_time)
```

```python
def search(file,pattern = r'\d{3}-\d{3}-\d{4}'):
    f = open(file,'r')
    text   = f.read()

    if re.search(pattern,text):
        print("string found")
        return re.findall(pattern,text  )
    else:
        return ''
```
