```python
work_hours = [('tom',2749),('jerry',2929),('peter',2439),('harry',2123),('ron',2789)]
```

```python
# Which employee has more working hours 

def employee_hrs(work_hours):
    curr_hrs = 0
    employee_of_month = ''
    for employees,hours in work_hours:
        if hours > curr_hrs:
            curr_hrs = hours
            employee_of_month = employees
        else:
            pass
    return (employee_of_month,curr_hrs)
```

```python
employee_hrs(work_hours)
```
